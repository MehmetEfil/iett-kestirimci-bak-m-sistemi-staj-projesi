```python
import pandas as pd

df = pd.read_csv("C:/Users/asus/Desktop/Datathon/panel_data/temiz_veri/ariza_model.csv", low_memory=False)

null_sayisi = df.isnull().sum()
null_var = null_sayisi[null_sayisi > 0]
null_oran = (null_var / len(df) * 100).round(1)

print(f"Toplam satır: {len(df):,}")
print(pd.DataFrame({'null_sayisi': null_var, 'oran_%': null_oran}))
```
